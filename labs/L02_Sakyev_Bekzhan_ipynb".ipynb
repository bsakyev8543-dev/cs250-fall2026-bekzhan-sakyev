```python
answer_1 = "7"
answer_2 = "34"
answer_3 = "12.0"
answer_4 = "error"
answer_5 = "False"

expected = {"answer_1": "7", "answer_2": "34", "answer_3": "12.0",
            "answer_4": "error", "answer_5": "False"}
given, missing = {}, []
for k in expected:
    v = globals().get(k)
    if v is None:
        missing.append(k)
    else:
        given[k] = str(v).strip().lower()

if missing:
    print("Not answered yet:", ", ".join(missing))
else:
    wrong = [k for k in expected if given[k] != expected[k].lower()]
    if wrong:
        print("Not quite - look again at:", ", ".join(wrong))
    else:
        print("PASSED - all five predictions are correct.")
```
